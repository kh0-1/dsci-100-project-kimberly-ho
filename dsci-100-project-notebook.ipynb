#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
tourney_calc <- basketball_data |>
                select(-tourney_finish) |>
                summarize(across(conf_wins:reg_wins, list(mean = mean, max = max, min = min))) # seed values excluded as they behave as discrete options despite being numbers 
tourney_calc
```

**Visualization 1**: Depicts relationship between a team's regional and conference win totals in a season. It is a weakly positive 
                 relationship, indicating that teams who do well in their regional matchups win more conference head-to-heads as
                 well. This is a way to mark the good teams, as they would have the highest win totals going into the tournament.
                 Also, this confirms that the two variables are not contradictory and can attest to the strength of a team, meaning
                 the teams with higher totals should be in a better position to do well in the tournament. 

```r
wins_plot <- basketball_data |>
            ggplot(aes(x = reg_wins, y = conf_wins)) +
            geom_point() +
            labs(x = "Regional Wins (Count)", y = "Conference Wins (Count)") +
            ggtitle("Relationship Between Regional Wins and Conference Wins") +
            theme(text = element_text(size = 20))
wins_plot
```

**Visualization 2**: Below are two visualizations that effectively show the same relationship using the original basketball data
                 frame. The first option shows the relationship between a team's seed and their tournament wins, a total which indicates how far into the tournament a team got. The second graph option shows the same relationship but separates into each round result, clarifying which seeds were able to get furthest in the tournaments. These visualizations show that in general, higher seeds (coming into the tournament with more wins) were more likely to do well, supporting the strength of the chosen variables in predicting the outcome class.

```r
options(repr.plot.width = 12, repr.plot.height = 8)
seed_wins_plot <- basketball |>
            ggplot(aes(x = seed, y = tourney_wins)) +
            geom_point() +
            labs(x = "Seed", y = "Tournament Wins (Count)") +
            ggtitle("Relationship Between Team Seed and Tournament Wins") +
            theme(text = element_text(size = 20))
seed_wins_plot
```

```r
options(repr.plot.width = 14, repr.plot.height = 20)
wins_split_plot <- basketball |>
            ggplot(aes(x = seed, y = tourney_wins)) +
            geom_point() +
            labs(x = "Seed Wins (Count)", y = "Tournaent Wins (Count)") +
            ggtitle("Relationship Between Seed and Tournament Wins at Different Rounds") +
            facet_grid(rows = vars(tourney_finish)) +
            theme(text = element_text(size = 20))
wins_split_plot
```
